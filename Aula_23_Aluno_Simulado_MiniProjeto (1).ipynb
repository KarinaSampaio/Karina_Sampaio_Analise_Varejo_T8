# SPRINT 2 — LIMPEZA E PADRONIZAÇÃO

| Problema | Decisão | Justificativa |

| Espaços e capitalização diferentes | Padronizar os textos | Reunir rótulos equivalentes |

# SPRINT 3 — DUAS FUNÇÕES

Crie:
1. uma função de tratamento;
2. uma função de resumo.

Ambas precisam ter parâmetro e `return`.

````python
# Primeira função de tratamento de dados

def tratar_dados(tabela):
  tratado = tabela.copy()

  for coluna in ["genero", "categoria", "cidade", "canal"]:
    tratado[coluna] = (
        tratado[coluna]
        .astype("string")
        .str.strip()
        .replace("", pd.NA)
        .str.title()
    )

  # Corrigido para usar dicionários no replace
  tratado["genero"] = (
      tratado["genero"]
      .replace("Não Informado", "Não informado")
      .fillna("Não informado")
  )

  tratado["cidade"] = tratado["cidade"].replace({
      "rio de Janeiro": "Rio de Janeiro",
      "SÃO PAULO": "São Paulo"
  })

  tratado["categoria"] = tratado["categoria"].replace({
      "ELETRÔNICOS": "Eletrônicos",
      "moda": "Moda"
  })

  # Retirar a duplicata
  tratado = tratado.drop_duplicates().copy()

  return tratado
````

````python
# função de valores

def gerar_resumo(tabela):
  valores = tabela["valor_compra"]

  return {
      "contagem_valida": valores.count(),
      "media": valores.mean(),
      "minimo": valores.min(),
      "maximo": valores.max()
  }
````

````python
df_limpo = tratar_dados(df_original)

resumo = gerar_resumo(df_limpo)
````

````python
# Conferindo tratamento antes da análise

print("Linhas antes: ", len(df_original))
print("Linhas depois: ", len(df_limpo))

print(
    "Duplicatas restantes: ",
    df_limpo.duplicated().sum()
)

display(df_limpo.isna().sum())

print(
    "Identificadores de clientes distintos: ",
    df_limpo["cliente_id"].nunique()
)
````

# SPRINT 4 — ANÁLISE

Faça:
- contagem válida, média, mínimo e máximo de uma coluna numérica adequada;
- um filtro;
- dois `groupby()` diferentes;
- interpretação dos resultados.

````python
# Representação das estatísticas

print("Valores válidos: ", resumo["contagem_valida"])
print("Média: ", round(resumo["media"], 2))
print("Mínimo: ", resumo["minimo"])
print("Máximo: ", resumo["maximo"])
````

Dos 48 registros tratados, 46 possuem valor de compra preenchido.

A média dos valores de compra conhecidos é de aproximadamente R$ 716,45

````python
# Compras acima do valor da média

registros_acima_media = df_limpo.loc[
    df_limpo["valor_compra"] > resumo["media"],
    ["registro_id", "cliente_id", "valor_compra"]
]

print("Quantidade de registros acima da média: ", len(registros_acima_media))

display(registros_acima_media)
````

````python
# Filtro de valores acima de 600

registros_acima_600 = df_limpo[
    df_limpo["valor_compra"] > 600

]

print(
    "Registros acima de R$ 600: ",
    len(registros_acima_600)
)

display(registros_acima_600)
````

Analisamos 22 registros acima da média de R$ 716,46.

Além disso, vimos a presença de 26 acima do valor de R$ 600.

````python
# Agrupamento por categoria

registros_por_categoria = (
    df_limpo.groupby("categoria")
    .size()
    .sort_values(ascending=False)
)

display(registros_por_categoria)
````

````python
# Agrupamento por canal

por_canal = (
    df_limpo.groupby("canal")
    .size()
    .sort_values(ascending=False)
)

display(por_canal)
````

````python
# Agrupar mais de uma coluna - canal e valor_compra

grupos_canal = (
    df_limpo.groupby("canal")["valor_compra"]

)

resumo_por_canal = pd.DataFrame({
    "registros": grupos_canal.size(),
    "valores_validos": grupos_canal.count(),
    "soma_valores_conhecidos": grupos_canal.sum(min_count=1)
})
display(resumo_por_canal.round(2))
````

````python
# Agrupamento de mais de uma coluna - categoria por valor_compra

grupos_categoria = (
    df_limpo.groupby("categoria")["valor_compra"]
    .sum()
    .sort_values(ascending=False)
)

display(grupos_categoria)

````

Moda e Eletrônicos, possuem a maior quantidade de registros por categoria. O App, possui o maior número de registros em canal.

App apesar de ter o maior número de registros, não possui a maior soma de valor de compra, essa sendo para Site, que possui a soma de valor de compra de R$ 11.660.

Já a categoria Moda, é a que possui o maior valor de compra, com aproximadamente R$ 7.620.
